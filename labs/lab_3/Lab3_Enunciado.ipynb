**Respuesta:**

````python
def to_snake_case(s: str) -> str:
    c = s.lower().replace(" ", "_").strip()  # el strip para quitar espacios a los bordes
    return c
````

````python
def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    # Cambiar nombre de las columnas a snake case
    df.columns = [to_snake_case(col) for col in df.columns]
    # eliminar palabra 'race'
    df.rename(columns={"race/ethnicity": "ethnicity"}, inplace=True)  # inplace cambia original y no una copia
    # eliminar duplicados
    df.drop_duplicates(keep="first", inplace=True)
    # forzar tipo de columna
    col_num_names = ["math_score", "reading_score", "writing_score"]
    for col in col_num_names:
        df.loc[:, col] = pd.to_numeric(df[col], errors="coerce")
    # eliminar nulos
    df = df.dropna(axis=0, how="any")
    # eliminar registro con notas anomalas
    df = df[(df["math_score"] >= 0) & (df["math_score"] <= 100)]
    df = df[(df["reading_score"] >= 0) & (df["reading_score"] <= 100)]
    df = df[(df["writing_score"] >= 0) & (df["writing_score"] <= 100)]

    return df
````

````python
df_cleaned = clean_data(df_grades)
df_cleaned
````

### 1.3.1 Transformación de notas a "Nota Chilena" 🇨🇱 [0.5 puntos]

En una de las reuniones con Don Caguayo, este les comenta su intención de pasar todas sus notas a la "Escala Chilena". Para esto les facilita la siguiente ecuación:

$$Nota \, Chilena= \% \, de \, logro \times 0.06 + 1$$

Utilizando la ecuación entregada, cree una función que le permita transformar las notas (que están en escala 0-100) a la escala chilena [1,7] y entregue las notas redondeadas al segundo decimal.

**To-Do**

- [ ] Implementa la función `to_chilean_grade` que toma un **DataFrame** y el nombre de una **columna** con notas en escala 0-100, y retorna una **Serie** con las notas transformadas a escala chilena (rango 1.0-7.0). [0.3 puntos]
- [ ] Agrega las nuevas series de notas de matemáticas (columna `math_score_cl`), lectura (columna `reading_score_cl`) y escritura (columna `writing_score_cl`) al dataframe `df_cleaned`. [0.3 puntos]

**Respuesta Código:**

````python
# En esta celda implementa la operación solicitada
def to_chilean_grade(df: pd.DataFrame, columna: str) -> tuple[pd.Series, pd.Series]:
    nota = df.loc[:, columna]
    nota_cl = (nota * 0.06 + 1).round(2)
    return nota, nota_cl
````

````python
# En esta celda ejecuta la función y luego usa indexadores para agregar las nuevas columnas al dataframe limpio
nota_math, nota_math_cl = to_chilean_grade(df_cleaned, "math_score")
nota_reading, nota_reading_cl = to_chilean_grade(df_cleaned, "reading_score")
nota_writing, nota_writing_cl = to_chilean_grade(df_cleaned, "writing_score")

df_cleaned.loc[:, "math_score_cl"] = nota_math_cl
df_cleaned.loc[:, "reading_score_cl"] = nota_reading_cl
df_cleaned.loc[:, "writing_score_cl"] = nota_writing_cl
````

### 1.3.2 Promedios 📋 [0.3 puntos]

Con todas las notas transformadas, la prestigiosa Universidad De la Cachaña necesita obtener el promedio general (GPA) de cada uno de los alumnos. Para esto se le solicita que calcule el promedio de las tres asignaturas y lo anexe como una nueva columna al DataFrame. 

Cabe señalar que las notas promedio deben ser **redondeadas** al segundo decimal.

**To-Do:**
- [ ] Implementa la función `calculate_average` que dado un dataframe con las notas por alumno, calcula y retorna dos **Series**: una con el promedio de las notas originales (escala 0-100) y otra con el promedio de las notas chilenas.
- [ ] Ejecuta la función y luego, usando indexadores apropiados, agrega las columnas `mean` (promedio escala 0-100) y `mean_cl` (promedio escala chilena) a `df_cleaned`.


**Respuesta:**

````python
# En esta celda implementa la operación solicitada
def calculate_average(df: pd.DataFrame) -> tuple[pd.Series, pd.Series]:
    cols_orig = ["math_score", "reading_score", "writing_score"]
    cols_cl = ["math_score_cl", "reading_score_cl", "writing_score_cl"]

    mean_score = df.loc[:, cols_orig].mean(axis=1).round(2)
    mean_score_cl = df.loc[:, cols_cl].mean(axis=1).round(2)

    return mean_score, mean_score_cl
````

````python
# En esta celda ejecuta la función y luego usa indexadores para agregar la nueva columna al dataframe limpio
mean, mean_cl = calculate_average(df_cleaned)

df_cleaned.loc[:, "mean"] = mean
df_cleaned.loc[:, "mean_cl"] = mean_cl
````

### 1.3.5 Padres Helicóptero 👨‍👩‍👧

Cree una función que:

1. Reciba el `DataFrame` y una lista con nombres de alumnos
2. Extraiga las notas chilenas de matemáticas, escritura, lectura y promedio general (`mean_cl`) de los alumnos listados en `student_list` [0.3 puntos]
3. Genere una nueva columna transitoria `is_failing` (**asegurarse de no agregarla al dataframe original**) cuyo valor sea `True` si el alumno está reprobando o `False` en caso contrario [0.3 puntos]
4. Ordene por promedio chileno, de mayor a menor [0.2 puntos]
5. Reinicie el índice para que este comience desde 0 [0.2 puntos]


**Hint:** Puede serle de utilidad utilizar `isin()`. Recuerde además usar una copia del dataframe original para no modificarlo con la nueva columna.

**Respuesta:**

````python
def query_students(df: pd.DataFrame, student_list: list[str]) -> pd.DataFrame:
    df_copy = df.copy()
    df_copy = df_copy.loc[df_copy["names"].isin(student_list)]  # seleccionamos solo los estudiantes de la lista
    cols_cl = ["names", "math_score_cl", "reading_score_cl", "writing_score_cl", "mean_cl"]
    df_copy = df_copy.loc[:, cols_cl]  # seleccionamos solo las columnas que piden
    df_copy.loc[:, "is_failing"] = df_copy["mean_cl"] < 4.0  # booleano
    df_copy = df_copy.sort_values(by="mean_cl", ascending=False)
    df_copy = df_copy.reset_index(drop=True)

    return df_copy
````

````python
# En esta celda implementa la operación solicitada
def get_categories(df: pd.DataFrame) -> tuple[pd.Series, pd.Series]:
    labels_cut = ["B1", "B2", "B3", "B4", "B5"]
    labels_qcut = ["Q1", "Q2", "Q3", "Q4", "Q5"]
    cut_cl = pd.cut(df["mean_cl"], bins=5, labels=labels_cut)
    qcut_cl = pd.qcut(df["mean_cl"], q=5, labels=labels_qcut)

    return cut_cl, qcut_cl
````

````python
# indexadores
cut_cl, qcut_cl = get_categories(df_cleaned)

df_cleaned.loc[:, "cut_cl"] = cut_cl
df_cleaned.loc[:, "qcut_cl"] = qcut_cl
````

**Respuesta 📝:**

```
La categoría 'cut' divide el rango del promedio de notas (1 al 7) en 5 intervalos, sin tomar la cantidad de estudiantes, sino solo sus notas. En cambio, qcut divide en cuantiles, es decir, para este caso toma el 20% de los alumnos (ordenados de menor a mayor) y le asigna el primer quintil y así sucesivamente; ordenando los alumnos por nota pero fijando los intervalos por posición. Es por esto que los grupos generados por qcut están balanceados en cantidad de datos, aunque los rangos de notas tengan "anchos" distintos.
```
