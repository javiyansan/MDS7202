**Respuesta Código:**

````python
# Primero deberan cargar los datos usando un lector apropiado
df_grades = pd.read_json("students_grades.json")
````

````python
def analyze_df(df: pd.DataFrame) -> None:
    dimensiones = df.shape
    columnas = df.columns
    primeros_5 = df.head(5)
    ultimos_5 = df.tail(5)
    random_5 = df.sample(5)
    nulos = df.isna().sum()
    unicos = df.nunique().sum()
    duplicados = df.duplicated().sum()
    df_dupli = df[df.duplicated()]
    # lo tenemos que pasar a numero ya que lo esta identificando como texto las variables numericas
    col_names = ["math score", "reading score", "writing score"]
    df[col_names] = df[col_names].apply(pd.to_numeric, errors="coerce")
    descrip_num = df[col_names].describe()

    print("Análisis del dataframe")
    print("===" * 5)
    # 1. Imprimir las dimensiones del dataframe
    print(f"Dimensiones (Filas, Columnas): {dimensiones}")
    print("---" * 10)
    # 2. Imprimir los nombres de las columnas
    print(f"\nColumnas presentes: {list(columnas)}")
    print("---" * 10)
    # 3. Imprimir los primeros 5, ultimos 5 y muestre aleatorio de 5.
    print("Primeras 5 filas")
    display(primeros_5)
    print("Últimas 5 filas")
    display(ultimos_5)
    print("Muestra Aleatoria de 5 filas")
    display(random_5)
    # 4. Descripción de los valores numericos
    print("Descripción de Variables Numericas")
    display(descrip_num)
    print("---" * 10)
    # 5. Cantidad de valores nulos
    print("Valores Nulos por Columna:")
    print(nulos)
    print("---" * 10)
    # 6. Cantidad de valores unico
    print(f"Cantidad de valores únicos: {unicos}")
    print("Cantidad de valores únicos por columna:")
    display(df.nunique())
    print("---" * 10)
    # 7. Total de filas con duplicados
    print(f"Cantidad de filas duplicadas: {duplicados}")
    print("---" * 10)
    # 8. Mostrar dataframe de filas duplicadas
    print("Filas duplicadas:")
    display(df_dupli)
    print("---" * 10)

    # 9. Histograma de cada tipo de nota
    histograma = px.histogram(df, x=col_names, title="Histograma Notas por Tipo")
    histograma.show()
````

````python
analyze_df(df_grades)
````
